### 2. Standardize Headers and Text Fields

#### Headers

```python
# Standardize Header Format
df_cqi.columns = (
    df_cqi.columns
    .str.strip()
    .str.lower()
    .str.replace('\n', ' ')
    .str.replace('\r', '')
    .str.replace('.', '_')
)
```

```python
# Work check
df_cqi.columns.tolist()
```

```python
# Rename headers to be consistant 
df_cqi.rename(columns={
    'country_of_origin': 'origin_country',
    'in_country_partner': 'local_partner'
}, inplace=True)
```

```python
# Work check
df_cqi.columns.tolist()
```

#### Standardize Text Field Values

```python
# ID text columns
text_columns = df_cqi.select_dtypes(include='object').columns.tolist()
```

```python
# Define Standardization Function
def standardize_text(text):
    if pd.isna(text):
        return text
    return (
        str(text)
        .strip()                     # Remove leading/trailing whitespace
        .lower()                     # Convert to lowercase
        .replace('\n', ' ')          # Remove line breaks
        .replace('\r', '')           # Remove carriage returns
    )
```

```python
# Apply to text columns
df_cqi[text_columns] = df_cqi[text_columns].applymap(standardize_text)
```

```python
# Work check
df_cqi[['species', 'origin_country', 'processing_method']].sample(5)
```

### 3. Normalize Inputs

```python
# Adressing 1. origin_country
# Normalize origin_country values
origin_country_map = {
    "cote d?ivoire": "côte d'ivoire",
    "tanzania, united republic of": "tanzania",
    "united states (hawaii)": "united states",
    "united states (puerto rico)": "puerto rico"
}

df_cqi['origin_country'] = df_cqi['origin_country'].replace(origin_country_map)
```

```python
# Work check
print(df_cqi['origin_country'].value_counts(dropna=False).sort_index())
```

### 4. Missing Values

```python
# Create a summary table of missing values for CQI dataset, excluding 0s
missing_count = df_cqi.isnull().sum()
missing_percent = (df_cqi.isnull().mean() * 100).round(2)

missing_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing %': missing_percent
})

# Filter out rows with 0 missing values
missing_summary = missing_summary[missing_summary['Missing Count'] > 0]

# Sorted by highest missing %
missing_summary = missing_summary.sort_values(by='Missing %', ascending=False)

# Display the summary
missing_summary
```

Several columns contain missing values that cannot be reliably imputed. To preserve transparency and analytical integrity:
- Numeric columns (altitude_mean_meters, harvest_year) will remain as NaN to avoid introducing bias or distortion.
- String columns (variety, processing_method, owner) will be filled with "Unknown" to maintain categorical structure while signaling absence.
- This approach ensures the dataset remains unskewed, auditable, and ready for downstream analysis.
- I will investigate region and origin_country further to determine whether missing values can be inferred based on known patterns or relationships.


```python
# Display row where origin_country is missing
missing_country_row = df_cqi[df_cqi['origin_country'].isna()]
print(missing_country_row)
```

After reviewing the country–region pairs, I’ve determined that I cannot safely impute region based on origin_country. Many countries have multiple distinct regions or highly specific entries, and forcing a single match would risk skewing the analysis. To preserve transparency and analytical integrity, missing region values will be filled with - unknown.


```python
# Define string columns to fill
string_cols = ['variety', 'processing_method', 'owner', 'region']

# Fill missing string values with "Unknown"
df_cqi[string_cols] = df_cqi[string_cols].fillna('unknown')
```
