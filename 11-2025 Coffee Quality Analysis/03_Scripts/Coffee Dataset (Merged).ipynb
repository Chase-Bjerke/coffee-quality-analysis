## 3. Merge

```python
# First rename column headers to match to make the merge easier. coffee_country & coffee_region will be the headers.
df_cqi.rename(columns={'origin_country': 'coffee_country', 'region': 'coffee_region'}, inplace=True)
```

```python
# Merge. Use left merge on the two PKs.
df_coffee_merged = pd.merge(df_top, df_cqi, on=['coffee_country', 'coffee_region'], how='left')
```

## 5. Merge Corrections

```python
# Patch Known Regions with Missing Countries
region_country_patch = {
    'hawaii': 'united states',
    'oromia': 'ethiopia',
    'yirgacheffe': 'ethiopia',
    'gedeo': 'ethiopia',
    'guji': 'ethiopia',
    'sumatra': 'indonesia',
    'valle del cauca': 'colombia'
}
```

```python
# Apply patch
df_top['coffee_country'] = df_top.apply(
    lambda row: region_country_patch.get(row['coffee_region'].lower(), row['coffee_country'])
    if row['coffee_country'].lower() == 'unknown' else row['coffee_country'],
    axis=1
)
```

```python
# Work check
df_top[df_top['coffee_country'].isin(['unknown', 'united states', 'ethiopia', 'indonesia', 'colombia'])][['coffee_country', 'coffee_region']].drop_duplicates().head(10)
```

```python
# Patch on region and country
region_country_patch = {
    'gedeo': 'guji',
    'guji': 'guji',
    'boquete': 'panama',
    'valle del cauca': 'cauca'  # or 'huila' depending on CQI listings
}
```

```python
df_top['coffee_country'] = df_top.apply(
    lambda row: region_country_patch.get(row['coffee_region'].lower(), row['coffee_country'])
    if row['coffee_country'].lower() == 'unknown' else row['coffee_country'],
    axis=1
)

df_top['coffee_region'] = df_top['coffee_region'].replace(region_country_patch)
```

```python
# patch region and country v2
region_country_patch.update({
    'volcan': 'panama',
    'panama': 'panama'  # for rows where country is blank
})
```

```python
df_top['coffee_country'] = df_top.apply(
    lambda row: region_country_patch.get(row['coffee_region'].lower(), row['coffee_country'])
    if row['coffee_country'].lower() == 'unknown' else row['coffee_country'],
    axis=1
)
```
